> 
> **Pearson's Correlation Coefficient ($r$)** measures the strength and direction of a **linear relationship** between two continuous variables:
> $$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$
> 
> **Intuition from the Matrices**:

>    - *Intuition*: Higher GDP per capita allows governments and citizens to afford superior healthcare, infrastructure, clean drinking water, and nutrition, increasing life expectancy. The correlation dramatically increased after 1952.
> 
> 2. **`pop` vs. `lifeExp` & `pop` vs. `gdpPercap` (Demographics vs. Health/Wealth)**:

**b.** Compute Spearman's Rank Correlation of the previous columns in 1952, 1987 and 2007. What do you observe?

```python
# 1.b Spearman's Rank Correlation for 1952, 1987, and 2007
for yr in years:
    sub_df = gapminder[gapminder['year'] == yr][quant_cols]
    corr_spearman = sub_df.corr(method='spearman')
    print(f"=== Spearman Rank Correlation Matrix ({yr}) ===")
    display(corr_spearman.round(4))
    print()
```

> ### **What Do You Observe? (Pearson vs. Spearman)**:
> 
> Comparing `lifeExp` vs `gdpPercap` side-by-side:
> - **1952**: Pearson $r = 0.278$ vs. **Spearman $\rho = 0.769$** ($+0.491$ difference!).
> - **1987**: Pearson $r = 0.750$ vs. **Spearman $\rho = 0.892$** ($+0.142$ difference).
> - **2007**: Pearson $r = 0.679$ vs. **Spearman $\rho = 0.857$** ($+0.178$ difference).
> 
> **Why is Spearman's correlation substantially higher than Pearson's, especially in 1952?**
> 1. **Spearman evaluates Monotonicity, Pearson evaluates Linearity**:
>    - Pearson tests whether points lie along a straight line ($y = ax + b$).
>    - Spearman ranks the values ($rank(x)$ vs. $rank(y)$) and evaluates whether $y$ increases whenever $x$ increases, regardless of whether the curve is linear or concave.
> 2. **Non-linear Relationship (The Preston Curve)**:
>    - The true relationship between income and life expectancy is **logarithmic** (diminishing marginal returns): going from \$500 to \$5,000 saves millions of lives; going from \$30,000 to \$35,000 yields only modest longevity gains.
> 3. **Immunity to Outliers**:
>    - Pearson is severely dragged down by extreme outliers in raw dollars (e.g. Kuwait in 1952). Spearman converts raw dollars into ranks ($1, 2, ..., 142$), completely neutralizing the leverage of extreme numerical outliers!

**d.** Revisit your intuition of the correlation matrix in year 1952 from question **(a)**, can you see why we observed such a (poor) correlation in 1952?!

```python
# Investigating the 1952 anomaly: The Kuwait Outlier
df1952 = gapminder[gapminder['year'] == 1952].copy()
top_gdp_1952 = df1952.sort_values(by='gdpPercap', ascending=False).head(5)

print("=== Top 5 Richest Countries in 1952 ===")
display(top_gdp_1952[['country', 'continent', 'gdpPercap', 'lifeExp', 'pop']])

# Correlation with Kuwait vs. without Kuwait
r_with = df1952['lifeExp'].corr(df1952['gdpPercap'], method='pearson')
df1952_no_kuwait = df1952[df1952['country'] != 'Kuwait']
r_without = df1952_no_kuwait['lifeExp'].corr(df1952_no_kuwait['gdpPercap'], method='pearson')

print(f"\n* 1952 Pearson r WITH Kuwait:    {r_with:.4f}")
print(f"* 1952 Pearson r WITHOUT Kuwait: {r_without:.4f}  (nearly DOUBLED!)")
print(f"* 1952 Pearson r on log(gdpPercap): {np.log(df1952['gdpPercap']).corr(df1952['lifeExp']):.4f}")
```

> ### **Why was the Pearson correlation so poor in 1952?**
> 
> **The Kuwait Oil Outlier!**
> 1. In **1952**, **Kuwait** had an astronomical GDP per capita of **\$108,382** due to early commercial oil discovery paired with a tiny population (~160,000 citizens).
> 2. The second richest country was Switzerland at only \$14,734—meaning Kuwait's GDP per capita was over **7 times higher** than the next wealthiest country!
> 3. Despite immense oil revenues, Kuwait's life expectancy in 1952 was only **55.57 years** (as modern healthcare infrastructure was only beginning to develop).
> 4. In a linear regression ($r$), this single point sits at the far extreme right of the plot ($x = 108,382$) with moderate $y = 55.6$. It acts as an **extreme leverage point**, pulling the entire linear regression line downward and flattening its slope!
> 5. Removing Kuwait alone causes the Pearson correlation in 1952 to leap from **$0.278$ to $0.531$**! Furthermore, applying a logarithmic transformation ($\log(\text{gdpPercap})$) raises the correlation to **$0.78$**.

> **Remark:** Pearson correlation matrix can summarize linear relationship between pairs of quantitative variables but it might be inaccurate and influenced by:
> - **Outliers** (extreme leverage points like Kuwait in 1952)
> - **Non-linearity** (logarithmic returns like the Preston Curve)
> - **Small sample size**
> - **Confounding (causal) variables** (continental healthcare infrastructure, geographical factors)

**Conclusion:** 
When analyzing variables that exhibit exponential growth or diminishing marginal returns (such as wealth vs. health), **Spearman's rank correlation** or **logarithmic transformations** provide a much truer and more robust representation of the underlying relationship than raw Pearson correlation.

**c.** We will try to confirm this using Analysis of Variance (ANOVA). 

In 2007,

- Perform one way ANOVA on `lifeExp` and `continent` (you can use `f_oneway` from `scipy.stats` module).

```python
# One-way ANOVA: lifeExp across continents in 2007
df2007 = gapminder[gapminder['year'] == 2007]

groups_lifeExp = [group['lifeExp'].values for name, group in df2007.groupby('continent')]
f_stat_life, p_val_life = stats.f_oneway(*groups_lifeExp)

print(f"=== One-Way ANOVA: `lifeExp` across continents (2007) ===")
print(f"F-statistic: {f_stat_life:.4f}")
print(f"p-value:     {p_val_life:.4e}")
if p_val_life < 0.05:
    print("Conclusion: Reject H0! There are statistically significant differences in mean life expectancy across continents.")
```

- Do the same with `gdpPercap` and `continent`.

```python
# One-way ANOVA: gdpPercap across continents in 2007
groups_gdp = [group['gdpPercap'].values for name, group in df2007.groupby('continent')]
f_stat_gdp, p_val_gdp = stats.f_oneway(*groups_gdp)

print(f"=== One-Way ANOVA: `gdpPercap` across continents (2007) ===")
print(f"F-statistic: {f_stat_gdp:.4f}")
print(f"p-value:     {p_val_gdp:.4e}")
if p_val_gdp < 0.05:
    print("Conclusion: Reject H0! There are statistically significant differences in mean GDP per capita across continents.")
```

```python
# Checking ANOVA Assumptions for Year 2007:
print("--- 1. Testing Homoscedasticity (Levene's Test for Equal Variances) ---")
lev_life = stats.levene(*groups_lifeExp)
lev_gdp = stats.levene(*groups_gdp)
print(f"lifeExp Levene test:  W = {lev_life.statistic:.4f}, p-value = {lev_life.pvalue:.4e} (Equal variances rejected!)")
print(f"gdpPercap Levene test: W = {lev_gdp.statistic:.4f}, p-value = {lev_gdp.pvalue:.4e} (Equal variances rejected!)")

print("\n--- 2. Testing Normality within Continents (Shapiro-Wilk Test) ---")
for cont, group in df2007.groupby('continent'):
    if len(group) >= 3:
        sh_life = stats.shapiro(group['lifeExp'])
        sh_gdp = stats.shapiro(group['gdpPercap'])
        print(f"{cont:10s} | lifeExp p={sh_life.pvalue:.4f} | gdpPercap p={sh_gdp.pvalue:.4e}")
```

- Are these results reliable? Why?

> **Your response**:
> **No, the classical One-Way ANOVA results are NOT strictly reliable.**
> 
> **Why?** One-Way ANOVA requires three foundational mathematical assumptions:
> 1. **Homoscedasticity (Homogeneity of Variances)**: All groups must have approximately equal variances ($\sigma_1^2 = \sigma_2^2 = ... = \sigma_k^2$).
>    - *Levene's test* yields $p = 4.9 	imes 10^{-5}$ for `lifeExp` and $p = 4.3 	imes 10^{-5}$ for `gdpPercap`, **severely rejecting equal variances**.
> 2. **Normality of Residuals**: The values within each continental group should be normally distributed.
>    - *Shapiro-Wilk tests* show significant departures from normality for almost all continents (e.g. GDP per capita in Africa has $p = 7.4 	imes 10^{-9}$ due to extreme skewness).
> 3. **Unequal Group Sizes**: Group sizes are highly unbalanced (Africa: 52, Americas: 25, Asia: 33, Europe: 30, Oceania: 2). When sample sizes and variances are both unequal, the classical $F$-test is sensitive to Type I and Type II errors.

- Propose ideas that might solve this problem or an alternative method.

> **Your response**:
> 1. **Logarithmic Transformation**: Apply a $\log$ transformation to `gdpPercap` ($\log_{10}(\text{gdpPercap})$). This dramatically reduces right-skewness, compresses outliers, and stabilizes group variances.
> 2. **Non-parametric Test (Kruskal-Wallis $H$-test)**: Use `scipy.stats.kruskal`, which evaluates differences in group ranks rather than means and requires **neither normality nor homoscedasticity**.
> 3. **Welch's ANOVA**: An adjusted ANOVA that explicitly relaxes the assumption of equal variances across groups.

```python
# Solution 1: Non-parametric Kruskal-Wallis H-test
kw_life = stats.kruskal(*groups_lifeExp)
kw_gdp = stats.kruskal(*groups_gdp)

print("=== Alternative: Kruskal-Wallis H-Test (Non-parametric) ===")
print(f"lifeExp:   H = {kw_life.statistic:.4f}, p-value = {kw_life.pvalue:.4e}")
print(f"gdpPercap: H = {kw_gdp.statistic:.4f}, p-value = {kw_gdp.pvalue:.4e}")
print("Both confirm massive, statistically significant continental differences even without normality!")

# Solution 2: Log transformation of GDP per capita + ANOVA
groups_log_gdp = [np.log10(group['gdpPercap'].values) for name, group in df2007.groupby('continent')]
f_log, p_log = stats.f_oneway(*groups_log_gdp)
print(f"\n=== ANOVA on log10(gdpPercap) ===")
print(f"F-statistic: {f_log:.4f}, p-value: {p_log:.4e}")
```

# **3. Quantitative vs Qualitative**

**a.** We have seen how life expectancy and economy vary across continents in 2007 ([Slide 25 of the course](https://hassothea.github.io/EDA_ITC/Slides/Bivariate_Analysis.html#/quan.-vs-qual.-3)). Compute $\eta$-squared correlation between `continent` and `lifeExp`, then `continent` with `gdpPercap` in 1952, 1987 and 2007. 

- Do you find the results reasonable?

```python
# 3.a Compute Eta-squared (correlation ratio) for continent vs. lifeExp and gdpPercap
def compute_eta_squared(df, categorical_col, numerical_col):
    """
    Computes Eta-squared (eta^2) = SS_between / SS_total
    """
    grand_mean = df[numerical_col].mean()
    ss_total = ((df[numerical_col] - grand_mean) ** 2).sum()
    
    group_stats = df.groupby(categorical_col)[numerical_col].agg(['count', 'mean'])
    ss_between = (group_stats['count'] * (group_stats['mean'] - grand_mean) ** 2).sum()
    
    return ss_between / ss_total

eta_results = []
for yr in [1952, 1987, 2007]:
    sub = gapminder[gapminder['year'] == yr]
    eta_life = compute_eta_squared(sub, 'continent', 'lifeExp')
    eta_gdp = compute_eta_squared(sub, 'continent', 'gdpPercap')
    eta_results.append({
        'Year': yr,
        'eta^2 (continent vs. lifeExp)': round(eta_life, 4),
        'eta^2 (continent vs. gdpPercap)': round(eta_gdp, 4)
    })

eta_df = pd.DataFrame(eta_results)
print("=== Eta-squared (Correlation Ratio) Summary ===")
display(eta_df)
```

> ### **Mathematical Definition & Interpretation of $\eta^2$**:
> 
> The **Correlation Ratio ($\eta^2$)** measures the proportion of total variance in a quantitative variable ($Y$) explained by a qualitative grouping variable ($X$):
> $$\eta^2 = \frac{SS_{\text{between}}}{SS_{\text{total}}} = \frac{\sum_{k} n_k (\bar{y}_k - \bar{y})^2}{\sum_{i} (y_i - \bar{y})^2}$$
> - Benchmark scale: $\eta^2 = 0.01$ (small effect), $\eta^2 = 0.06$ (medium effect), $\eta^2 \ge 0.14$ (large effect).
> 
> ### **Are the results reasonable?**
> **Yes, completely reasonable and insightful!**
> 1. **`continent` vs. `lifeExp`**:
>    - $\eta^2 \approx 0.650$ (1952), $0.592$ (1987), and $0.635$ (2007).
>    - This indicates that **around 60% to 65% of the global variation in life expectancy is explained solely by which continent a country belongs to!** Geographical healthcare disparities and systemic regional living standards create persistent continental clusters.
> 2. **`continent` vs. `gdpPercap`**:
>    - In **1952**: $\eta^2 = 0.048$ (very small!).
>      *Reason*: The oil outlier Kuwait was located in Asia with \$108,000+ GDP/capita, creating massive *within-continent* variance in Asia that completely dwarfed between-continent differences!
>    - In **1987 & 2007**: $\eta^2$ jumped to **$0.471$ and $0.424$** (large effect!).
>      *Reason*: As world economies diversified and global industrialization polarized Europe/Oceania vs. Africa/Asia, continental grouping became a powerful explanatory factor explaining over 42% of economic variance.

```python
# 4.a Evolution of all 4 correlation metrics from 1952 to 2007
time_records = []
all_years = sorted(gapminder['year'].unique())

for yr in all_years:
    sub = gapminder[gapminder['year'] == yr]
    r_pearson = sub['lifeExp'].corr(sub['gdpPercap'], method='pearson')
    r_spearman = sub['lifeExp'].corr(sub['gdpPercap'], method='spearman')
    eta_life = compute_eta_squared(sub, 'continent', 'lifeExp')
    eta_gdp = compute_eta_squared(sub, 'continent', 'gdpPercap')
    
    time_records.append({
        'year': yr,
        'Pearson (lifeExp ~ gdpPercap)': r_pearson,
        'Spearman (lifeExp ~ gdpPercap)': r_spearman,
        'eta2 (continent ~ lifeExp)': eta_life,
        'eta2 (continent ~ gdpPercap)': eta_gdp
    })

evol_df = pd.DataFrame(time_records)

# Plotting with Matplotlib & Seaborn
plt.figure(figsize=(12, 6))
plt.plot(evol_df['year'], evol_df['Pearson (lifeExp ~ gdpPercap)'], marker='o', linewidth=2.5, label='Pearson (lifeExp ~ gdpPercap)', color='#1f77b4')
plt.plot(evol_df['year'], evol_df['Spearman (lifeExp ~ gdpPercap)'], marker='s', linewidth=2.5, label='Spearman (lifeExp ~ gdpPercap)', color='#2ca02c')
plt.plot(evol_df['year'], evol_df['eta2 (continent ~ lifeExp)'], marker='^', linewidth=2.5, label=r'$\eta^2$ (continent ~ lifeExp)', color='#ff7f0e')
plt.plot(evol_df['year'], evol_df['eta2 (continent ~ gdpPercap)'], marker='d', linewidth=2.5, label=r'$\eta^2$ (continent ~ gdpPercap)', color='#d62728')

plt.title('Evolution of Bivariate Correlation Metrics (1952 - 2007)', fontsize=15, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Correlation / Association Coefficient', fontsize=12)
plt.xticks(all_years, rotation=45)
plt.ylim(0, 1.0)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11, loc='center right')
plt.tight_layout()
plt.show()
```

> ### **Synthesis: What We Observed About the World from 1952 to 2007**:
> 
> 1. **Global Health Convergence (The Preston Curve Shift)**:
>    - Across the globe, **life expectancy rose dramatically** from an average of ~49 years in 1952 to over ~67 years in 2007.
>    - Importantly, the curve between GDP per capita and life expectancy shifted **upward over time**: even poor countries in 2007 achieved higher life expectancy than countries with identical wealth in 1952, thanks to the global diffusion of vaccines, oral rehydration therapy, clean water infrastructure, and antibiotics.
> 
> 2. **Economic Divergence**:
>    - While health converged, **economic inequality widened**. North America, Western Europe, and Oceania experienced explosive GDP per capita growth, while Sub-Saharan Africa remained stagnant near the poverty line.
>    - $\eta^2$ for continent vs. GDP per capita surged from $\approx 0.05$ to $> 0.42$, confirming that global wealth became starkly divided along continental lines.
> 
> 3. **The Rise of Asia**:
>    - Asia transformed from a continent where over 80% of countries were "unhealthy" in 1952 to an economic and health powerhouse by 2007, led by Japan, South Korea, Taiwan, Singapore, and the massive demographic and economic ascent of China and India.
> 
> 4. **Tragedy and Resilience (Cambodia vs. Thailand)**:
>    - Thailand pursued peaceful industrialization and tourism, achieving steady, uninterrupted growth in health (reaching 70.6 yrs) and wealth (>\$7,400).
>    - Cambodia suffered devastating historical rupture during the 1970s Khmer Rouge genocide, where life expectancy plunged to an unprecedented 31.2 years in 1977. However, the post-1980 era demonstrated astonishing human resilience, with life expectancy recovering back to ~60 years by 2007 and economic growth rekindling.
